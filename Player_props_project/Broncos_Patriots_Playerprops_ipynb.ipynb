# Broncos vs Patriots — Player Prop Analysis

This notebook logs pregame player prop assumptions, market movement,
and postgame outcomes for evaluation.

All baseline entries are frozen prior to kickoff.


## Objectives

- Record baseline player prop lines with timestamps
- Track pregame market movement and adjustments
- Preserve reasoning without hindsight bias
- Evaluate outcomes postgame
- Measure accuracy of predictions vs final stats

## Methodology

This notebook separates:
1. **Baseline assumptions** – lines and probabilities before market moves
2. **Pregame adjustments** – updated market trends, value shifts, and thesis support
3. **Postgame outcomes** – final stats, calculated lean, thesis validation, and notes

**No baseline edits are allowed once the week’s pregame snapshot is frozen.**

```python
from tabulate import tabulate

def display_table():
    print(tabulate(df, headers='keys', tablefmt='grid'))

# Display current pregame props
display_table()
```

```python
def update_adjustment(player, new_value, reason, prop=None):
    """Update Adjusted Value and pregame rationale for a player or specific prop"""
    mask = df['Player'] == player
    if prop:
        mask &= df['Prop'] == prop
    df.loc[mask, 'Adjusted Value'] = new_value
    df.loc[mask, 'Postgame Notes'] = df.loc[mask, 'Postgame Notes'].fillna('') + " | " + reason
    df.loc[mask, 'Current Timestamp'] = datetime.now().strftime("%Y-%m-%d %H:%M")

def record_outcome(player, final_stat, prop=None):
    if prop:
        mask = (df['Player'] == player) & (df['Prop'] == prop)
    else:
        mask = (df['Player'] == player)
    df.loc[mask, 'Final Stat'] = final_stat
    df.loc[mask, 'Result'] = "OVER" if final_stat > df.loc[mask, 'Baseline Prop'].values[0] else "UNDER"
    # Thesis validation placeholder
    df.loc[mask, 'Thesis Validation'] = df.loc[mask, 'Result']  # initially, can compare with Expected Outcome
```

```python
# ================================
# Step 7: Final Picks Summary
# ================================

print("FINAL PICKS SUMMARY")
print("-" * 60)

for _, row in df.iterrows():
    lean = row['Type']  # OVER or UNDER
    adjusted = row['Adjusted Value'] if row['Adjusted Value'] else row['Baseline Prop']
    confidence = row['Confidence'] if row['Confidence'] else "N/A"
    print(
        f"{row['Player']} — {row['Prop']}: "
        f"{lean} (Adjusted: {adjusted}, Confidence: {confidence})"
    )
```
