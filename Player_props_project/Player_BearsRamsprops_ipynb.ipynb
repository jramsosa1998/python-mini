```python
# ================================
# Step 1: Import Libraries
# ================================
import pandas as pd
from datetime import datetime
from tabulate import tabulate

# ================================
# Step 2: Define Baseline Player Props
# ================================
data = [
    {"Player": "Swift", "Prop": "Rushing Yards", "Baseline Prop": 54.5, "Current Prop": 55.5,
     "Baseline %": 60, "Date Logged (Baseline)": "2026-01-14 09:15",
     "Adjusted %": 42, "Date Logged (Adjustment)": "2026-01-16 10:30",
     "Rationale": "Rams upgraded run defense, Bears LT injured, running game plateau", "Outcome": None},
    
    {"Player": "Loveland", "Prop": "Receptions", "Baseline Prop": 5.5, "Current Prop": 5.5,
     "Baseline %": 68, "Date Logged (Baseline)": "2026-01-14 09:20",
     "Adjusted %": 45, "Date Logged (Adjustment)": "2026-01-16 10:35",
     "Rationale": "Volume suppression, efficiency > volume, play-action exploited by Rams", "Outcome": None},
    
    {"Player": "Loveland", "Prop": "Receiving Yards", "Baseline Prop": 56.5, "Current Prop": 56.5,
     "Baseline %": 65, "Date Logged (Baseline)": "2026-01-14 09:20",
     "Adjusted %": 65, "Date Logged (Adjustment)": "2026-01-16 10:35",
     "Rationale": "Efficiency plays preserved yardage even with reduced volume", "Outcome": None},
    
    {"Player": "Puka", "Prop": "Receptions", "Baseline Prop": 7.5, "Current Prop": 7.0,
     "Baseline %": 70, "Date Logged (Baseline)": "2026-01-14 09:25",
     "Adjusted %": 42, "Date Logged (Adjustment)": "2026-01-16 10:40",
     "Rationale": "Bears bracket Puka, Adams siphons targets, volume capped", "Outcome": None}
]

# ================================
# Step 3: Create DataFrame & Calculate Lean
# ================================
df = pd.DataFrame(data)
df["Lean"] = df["Adjusted %"].apply(lambda x: "OVER" if x > 50 else "UNDER")

# ================================
# Step 4: Function to Display Table
# ================================
def display_table():
    print(tabulate(df, headers='keys', tablefmt='grid'))

# ================================
# Step 5: Display the Table (Run this cell to see your baseline table)
# ================================
display_table()
```

```python
from datetime import datetime

def update_adjustment(player_name, new_prob, new_rationale):
    df.loc[df['Player'] == player_name, 'Adjusted %'] = new_prob
    df.loc[df['Player'] == player_name, 'Rationale'] += " | " + new_rationale
    df.loc[df['Player'] == player_name, 'Date Logged (Adjustment)'] = datetime.now().strftime("%Y-%m-%d %H:%M")
    df.loc[df['Player'] == player_name, 'Lean'] = "OVER" if new_prob > 50 else "UNDER"


def record_outcome(player_name, outcome):
    df.loc[df['Player'] == player_name, 'Outcome'] = outcome


def calculate_accuracy():
    valid = df[df['Outcome'].notnull()]
    if len(valid) == 0:
        return None
    return (valid['Lean'] == valid['Outcome']).mean() * 100
```
