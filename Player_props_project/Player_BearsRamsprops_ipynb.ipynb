# Bears vs Rams Player Prop Analysis

This notebook tracks player prop predictions for the Bears vs Rams matchup.

**Objectives:**
- Track baseline props from sportsbooks
- Adjust probabilities based on injuries, matchups, and trends
- Compare predictions to actual outcomes
- Calculate prediction accuracy

**Baseline Player Prop Analysis & Leans:**
- **Swift – Rushing Yards:** Expect **UNDER** due to Rams’ strong run defense and Bears’ LT injury.
- **Loveland – Receptions:** Expect **UNDER**, but **Receiving Yards** could increase because of efficiency and explosive plays.
- **Puka – Receptions:** Expect **UNDER** due to tight coverage and Adams siphoning targets.

> Note: Small line discrepancies (like 0.5 yards) are documented for transparency but do not affect the overall analysis.


```python
# ================================
# Step 1: Import Libraries
# ================================
import pandas as pd
from datetime import datetime
from tabulate import tabulate

# ================================
# Step 2: Define Baseline Player Props
# ================================
data = [
    {"Player": "Swift", "Prop": "Rushing Yards", "Baseline Prop": 54.5, "Current Prop": 55.5,
     "Baseline %": 60, "Date Logged (Baseline)": "2026-01-14 09:15",
     "Adjusted %": 42, "Date Logged (Adjustment)": "2026-01-16 10:30",
     "Rationale": "Rams upgraded run defense, Bears LT injured, running game plateau", "Outcome": None},
    
    {"Player": "Loveland", "Prop": "Receptions", "Baseline Prop": 5.5, "Current Prop": 5.5,
     "Baseline %": 68, "Date Logged (Baseline)": "2026-01-14 09:20",
     "Adjusted %": 45, "Date Logged (Adjustment)": "2026-01-16 10:35",
     "Rationale": "Volume suppression, efficiency > volume, play-action exploited by Rams", "Outcome": None},
    
    {"Player": "Loveland", "Prop": "Receiving Yards", "Baseline Prop": 56.5, "Current Prop": 56.5,
     "Baseline %": 65, "Date Logged (Baseline)": "2026-01-14 09:20",
     "Adjusted %": 65, "Date Logged (Adjustment)": "2026-01-16 10:35",
     "Rationale": "Efficiency plays preserved yardage even with reduced volume", "Outcome": None},
    
    {"Player": "Puka", "Prop": "Receptions", "Baseline Prop": 7.5, "Current Prop": 7.0,
     "Baseline %": 70, "Date Logged (Baseline)": "2026-01-14 09:25",
     "Adjusted %": 42, "Date Logged (Adjustment)": "2026-01-16 10:40",
     "Rationale": "Bears bracket Puka, Adams siphons targets, volume capped", "Outcome": None}
]

# ================================
# Step 3: Create DataFrame & Calculate Lean
# ================================
df = pd.DataFrame(data)
df["Lean"] = df["Adjusted %"].apply(lambda x: "OVER" if x > 50 else "UNDER")

# ================================
# Step 4: Function to Display Table
# ================================
def display_table():
    print(tabulate(df, headers='keys', tablefmt='grid'))

# ================================
# Step 5: Display the Table (Run this cell to see your baseline table)
# ================================
display_table()
```

### ⚠️ Line Logging Clarification — Loveland Receiving Yards

During early-week analysis (Jan 14), Loveland’s receiving yards were available at **55.5** at several sportsbooks.  
This line aligned with my thesis that he could exceed his yardage total through **efficiency rather than volume**, even with fewer expected targets.

When formalizing the tracking table, the receiving yards line was logged at **56.5**, representing the upper end of the market at the time.  
This resulted in a minor **0.5-yard discrepancy** relative to the initial evaluation.

**Key takeaway:**  
- **Receptions** were modeled as an **UNDER** due to volume suppression.  
- **Receiving yards** were modeled as an **OVER** due to efficiency on play-action and chunk plays.  

> This clarification preserves transparency around line variance and ensures accurate tracking without changing original timestamps or rationale.


```python
from datetime import datetime

def update_adjustment(player, new_prob, reason, prop=None):
    """Update Adjusted % and Lean for a single prop or all props for a player."""
    mask = (df['Player'] == player)
    if prop:
        mask &= (df['Prop'] == prop)
    df.loc[mask, 'Adjusted %'] = new_prob
    df.loc[mask, 'Rationale'] += " | " + reason
    df.loc[mask, 'Date Logged (Adjustment)'] = datetime.now().strftime("%Y-%m-%d %H:%M")
    df.loc[mask, 'Lean'] = "OVER" if new_prob > 50 else "UNDER"

    

def record_outcome(player_name, outcome, prop=None):
    if prop:
        df.loc[(df['Player'] == player_name) & (df['Prop'] == prop), 'Outcome'] = outcome
    else:
        df.loc[df['Player'] == player_name, 'Outcome'] = outcome


def calculate_accuracy():
    valid = df[df['Outcome'].notnull()]
    if len(valid) == 0:
        return None
    return (valid['Lean'] == valid['Outcome']).mean() * 100
```

## Pre-Game Model Snapshot (Before Final Injury / Line Movement)

This table represents my baseline probabilities and intuition-based adjustments
**before** late-week injury reports and market line movement.

Goal:
- Capture my reasoning at time T
- Compare against final outcomes
- Measure accuracy, not perfection


```python
# ================================
# Step 6: Apply Adjustments
# ================================

update_adjustment("Swift", 40, "Latest injury news lowers chance", prop="Rushing Yards")  
update_adjustment("Loveland", 45, "Volume suppressed due to Rams coverage", prop="Receptions")  
update_adjustment("Puka", 42, "Bears bracket Puka, Adams siphons targets", prop="Receptions")  
df.loc[(df['Player'] == "Puka") & (df['Prop'] == "Receptions"), 'Rationale'] = "Bears bracket Puka, Adams siphons targets"  

display_table()
```

### 🏈 Final Picks Summary

This cell prints a **clean, readable summary of each player's prop** showing:
- The **Lean** (OVER or UNDER) based on adjusted probabilities  
- The **Adjusted %** after pregame trend adjustments  

This allows quick comparison with **actual outcomes** later in the notebook.

```python
# ================================
# Step 7: Final Picks Summary
# ================================

print("FINAL PICKS SUMMARY")
print("-" * 30)

for _, row in df.iterrows():
    print(
        f"{row['Player']} {row['Prop']}: "
        f"{row['Lean']} (Adjusted {row['Adjusted %']}%)"
    )
```

### 🏈 Record Actual Outcomes & Calculate Accuracy

This cell records the **actual stats** from the **Bears vs Rams Divisional Playoff game on January 18, 2026** for each player prop.  

It calculates:
- Whether the **predictions were OVER or UNDER**  
- The **accuracy** of my pregame lean predictions  

This allows comparison of **pregame adjusted expectations** to the **final outcomes** in a high-stakes divisional playoff context.


```python
# ================================
# Step 8: Record Actual Outcomes
# ================================
record_outcome("Swift", 76, prop="Rushing Yards")           # Swift rushing yards
record_outcome("Loveland", 4, prop="Receptions")           # Loveland receptions
record_outcome("Loveland", 56, prop="Receiving Yards")     # Loveland receiving yards
record_outcome("Puka", 5, prop="Receptions")               # Puka receptions

# ================================
# Step 9: Calculate Actual Result (OVER/UNDER)
# ================================
df['Result'] = df.apply(lambda row: "OVER" if row['Outcome'] > row['Current Prop'] else "UNDER", axis=1)

# ================================
# Step 10: Calculate Accuracy of Lean Predictions
# ================================
accuracy = (df['Lean'] == df['Result']).mean() * 100
print(f"Prediction Accuracy: {accuracy:.2f}%")

# ================================
# Step 11: Display Updated Table
# ================================
display_table()
```
