## Objectives

## Methodology

This notebook separates the analysis into three structured phases, building on lessons learned from prior projects:

1. **Baseline assumptions** – capture player prop lines and probabilities before any market movement. These reflect initial expectations grounded in matchup, player usage, and historical trends.
2. **Pregame adjustments** – track updated market trends, shifts in value, and thesis support. Adjustments incorporate new information such as injuries, team strategy, or emerging tendencies.
3. **Postgame outcomes** – record final stats, compare actual performance against baseline and adjusted expectations, validate the thesis, and document key observations.

**Note:** Baseline entries are frozen once the pregame snapshot is captured to maintain process integrity and prevent hindsight bias.

**Project Reflection:** This version builds on the Bears vs Rams analysis by emphasizing structured reasoning, clearer adjustment tracking, and enhanced postgame reflection — demonstrating iterative improvement in predictive rigor and notebook organization.

This table captures the full lifecycle of each prop:

- Market alignment or disagreement

```python
# Step 4: Adjustment Function
def update_adjustment(player, new_value, reason, prop=None):
    """Update Adjusted Value and pregame rationale for a player or specific prop"""
    mask = df['Player'] == player
    if prop:
        mask &= df['Prop'] == prop
    df.loc[mask, 'Adjusted Value'] = new_value
    df.loc[mask, 'Postgame Notes'] = df.loc[mask, 'Postgame Notes'].fillna('') + " | " + reason
    df.loc[mask, 'Current Timestamp'] = datetime.now().strftime("%Y-%m-%d %H:%M")
```

```python
# Step 5: Record Final Stats Function
def record_final_stats(player, final_stat, prop=None):
    """Record final outcome and calculate Result / Thesis Validation"""
    if prop:
        mask = (df['Player'] == player) & (df['Prop'] == prop)
    else:
        mask = (df['Player'] == player)
    df.loc[mask, 'Final Stat'] = final_stat
    df.loc[mask, 'Result'] = "OVER" if final_stat > float(str(df.loc[mask, 'Baseline Prop'].values[0]).split()[1]) else "UNDER"
    # Thesis validation placeholder: compare Result to expected prop type
    df.loc[mask, 'Thesis Validation'] = "Correct" if df.loc[mask, 'Result'].values[0] == df.loc[mask, 'Type'].values[0] else "Wrong"
```

```python
# Step 6: Final Picks Summary
print("FINAL PICKS SUMMARY")
print("-" * 80)

for _, row in df.iterrows():
    lean = row['Type']
    adjusted = row['Adjusted Value'] if row['Adjusted Value'] else row['Baseline Prop']
    confidence = row['Confidence'] if row['Confidence'] else "N/A"
    final_stat = row['Final Stat'] if row['Final Stat'] else "N/A"
    result = row['Result'] if row['Result'] else "N/A"
    thesis_val = row['Thesis Validation'] if row['Thesis Validation'] else "N/A"
    
    print(
        f"{row['Player']} — {row['Prop']}: "
        f"{lean} (Adjusted: {adjusted}, Confidence: {confidence}) | "
        f"Final Stat: {final_stat} | Result: {result} | Thesis Validation: {thesis_val}"
    )
```
